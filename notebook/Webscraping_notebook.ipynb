### b. Now, use what you did for the first title, but extract the job title for all jobs on this page. Store the results in a list.  

```python
job_title_tags = soup.find_all('h2')
```

```python
job_titles = [x.text for x in job_title_tags]
```

```python
job_titles
```

### c. Finally, extract the companies, locations, and posting dates for each job. For example, the first job has a company of "Payne, Roberts and Davis", a location of "Stewartbury, AA", and a posting date of "2021-04-08".
```python
companies_all = soup.find_all('h3')
```

```python
companies = [x.text for x in companies_all]
```

```python
companies
```

```python
locations_all = soup.find_all('p', class_= 'location')
```

```python
locations_all
```

```python
locations = [x.text for x in locations_all]
```

```python
posting_date_all = soup.find_all('time')
```

```python
posting_date = [x.text for x in posting_date_all]
```
